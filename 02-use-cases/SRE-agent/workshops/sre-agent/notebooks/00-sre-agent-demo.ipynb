````python
# Create a simple demo backend server
demo_server_code = '''
import json
from http.server import HTTPServer, BaseHTTPRequestHandler
from urllib.parse import urlparse, parse_qs
import threading

# Simulated infrastructure data
DEMO_DATA = {
    "pods": {
        "pods": [
            {
                "name": "payment-service-7b9c4d-x5m1q",
                "namespace": "production", 
                "status": "CrashLoopBackOff",
                "restarts": 15,
                "memory_usage": "98%",
                "cpu_usage": "25%",
                "node": "worker-node-2",
                "age": "2h"
            },
            {
                "name": "payment-service-8c1d5e-y6n2r", 
                "namespace": "production",
                "status": "CrashLoopBackOff",
                "restarts": 12,
                "memory_usage": "95%",
                "cpu_usage": "30%",
                "node": "worker-node-1", 
                "age": "2h"
            }
        ]
    },
    "logs": {
        "logs": [
            {
                "timestamp": "2024-01-15T14:24:30.789Z",
                "service": "payment-service",
                "level": "ERROR", 
                "message": "java.lang.OutOfMemoryError: Java heap space at PaymentProcessor.processPayment(PaymentProcessor.java:87)"
            },
            {
                "timestamp": "2024-01-15T14:25:10.123Z",
                "service": "payment-service",
                "level": "CRITICAL",
                "message": "OutOfMemoryError: Cannot allocate 256MB for transaction buffer"
            },
            {
                "timestamp": "2024-01-15T14:25:45.456Z", 
                "service": "payment-service",
                "level": "ERROR",
                "message": "Failed to process payment for order #12345: Service temporarily unavailable"
            }
        ]
    },
    "metrics": {
        "performance": {
            "service": "payment-service",
            "response_time_avg": "8500ms",
            "response_time_p99": "15000ms", 
            "error_rate": "87%",
            "throughput": "2.3 req/sec",
            "success_rate": "13%",
            "memory_usage": "96%",
            "cpu_usage": "28%"
        }
    },
    "runbooks": {
        "procedures": [
            {
                "title": "OutOfMemoryError Resolution",
                "severity": "HIGH", 
                "steps": [
                    "1. Check JVM heap size configuration",
                    "2. Analyze memory usage patterns in APM", 
                    "3. Increase memory limits in deployment.yaml",
                    "4. Restart affected pods",
                    "5. Monitor for memory leak patterns"
                ],
                "escalation": "If issue persists after memory increase, escalate to Platform Engineering team"
            }
        ]
    }
}

class DemoHandler(BaseHTTPRequestHandler):
    def do_GET(self):
        path = urlparse(self.path).path
        
        if path == "/pods":
            self.send_response(200)
            self.send_header('Content-type', 'application/json')
            self.end_headers()
            self.wfile.write(json.dumps(DEMO_DATA["pods"]).encode())
        elif path == "/logs":
            self.send_response(200)
            self.send_header('Content-type', 'application/json')
            self.end_headers()
            self.wfile.write(json.dumps(DEMO_DATA["logs"]).encode())
        elif path == "/metrics":
            self.send_response(200)
            self.send_header('Content-type', 'application/json') 
            self.end_headers()
            self.wfile.write(json.dumps(DEMO_DATA["metrics"]).encode())
        elif path == "/runbooks":
            self.send_response(200)
            self.send_header('Content-type', 'application/json')
            self.end_headers()
            self.wfile.write(json.dumps(DEMO_DATA["runbooks"]).encode())
        else:
            self.send_response(404)
            self.end_headers()
    
    def log_message(self, format, *args):
        pass  # Suppress server logs

def start_server():
    server = HTTPServer(('localhost', 8080), DemoHandler)
    server.serve_forever()

if __name__ == "__main__":
    start_server()
'''

# Save demo server code
with open('demo_server.py', 'w') as f:
    f.write(demo_server_code)

print("📝 Demo server code created")
print("   File: demo_server.py")
print("   Ready to start simulated infrastructure")
````

## Part 3: AI Agent Investigation (10 minutes)

Now watch our AI SRE agent automatically investigate the problem.

````python
# Create SRE Tools using Strands @tool decorator

@tool
def query_kubernetes_api():
    """
    Query Kubernetes API to get pod status, deployments, and cluster events.
    
    Returns comprehensive Kubernetes cluster information including:
    - Pod statuses and resource usage
    - Container restart counts
    - Node assignments and health
    """
    try:
        response = requests.get(f"{DEMO_BACKEND_URL}/pods", timeout=5)
        pods_data = response.json()
        
        summary = f"Found {len(pods_data['pods'])} pods in cluster:\n"
        for pod in pods_data['pods']:
            summary += f"- {pod['name']}: {pod['status']} ({pod['restarts']} restarts, Memory: {pod['memory_usage']}, CPU: {pod['cpu_usage']})\n"
        
        return summary
    except Exception as e:
        return f"Error querying Kubernetes API: {e}"

@tool 
def search_application_logs():
    """
    Search application logs for errors, exceptions, and critical events.
    
    Analyzes recent log entries to identify:
    - Error patterns and exception stack traces
    - Critical system messages
    - Service failures and timeouts
    """
    try:
        response = requests.get(f"{DEMO_BACKEND_URL}/logs", timeout=5)
        logs_data = response.json()
        
        summary = f"Analyzed {len(logs_data['logs'])} recent log entries:\n"
        for log in logs_data['logs']:
            summary += f"- [{log['timestamp']}] {log['level']}: {log['message']}\n"
            
        return summary
    except Exception as e:
        return f"Error searching application logs: {e}"

@tool
def get_performance_metrics():
    """
    Retrieve performance metrics including response times, error rates, and throughput.
    
    Provides real-time metrics for:
    - Response time percentiles (avg, p99)
    - Error rates and success rates  
    - Service throughput and capacity utilization
    """
    try:
        response = requests.get(f"{DEMO_BACKEND_URL}/metrics", timeout=5)
        metrics_data = response.json()
        
        perf = metrics_data['performance']
        summary = f"Performance Metrics for {perf['service']}:\n"
        summary += f"- Response Time: {perf['response_time_avg']} (p99: {perf['response_time_p99']})\n"
        summary += f"- Error Rate: {perf['error_rate']} (Success: {perf['success_rate']})\n" 
        summary += f"- Throughput: {perf['throughput']}\n"
        summary += f"- Resource Usage: Memory {perf['memory_usage']}, CPU {perf['cpu_usage']}\n"
        
        return summary
    except Exception as e:
        return f"Error retrieving performance metrics: {e}"

@tool
def consult_runbooks():
    """
    Access operational runbooks and procedures for incident resolution.
    
    Provides access to:
    - Step-by-step resolution procedures
    - Escalation guidelines and contacts
    - Best practices for common issues
    """
    try:
        response = requests.get(f"{DEMO_BACKEND_URL}/runbooks", timeout=5)
        runbooks_data = response.json()
        
        summary = f"Available Operational Procedures:\n"
        for procedure in runbooks_data['procedures']:
            summary += f"\n{procedure['title']} (Severity: {procedure['severity']}):\n"
            for i, step in enumerate(procedure['steps'], 1):
                summary += f"  {step}\n"
            summary += f"Escalation: {procedure['escalation']}\n"
            
        return summary
    except Exception as e:
        return f"Error consulting runbooks: {e}"

# Create the SRE Agent with Strands
sre_system_prompt = """You are an expert Site Reliability Engineer (SRE) specializing in production incident response and troubleshooting.

Your role is to:
1. Investigate infrastructure problems systematically using available tools
2. Correlate data across Kubernetes, logs, metrics, and operational procedures  
3. Identify root causes with high confidence
4. Provide specific, actionable resolution steps
5. Suggest preventive measures to avoid future incidents

When investigating an incident:
- Start by gathering data from all relevant systems simultaneously
- Look for patterns and correlations across different data sources
- Focus on technical accuracy and provide confidence levels
- Generate specific recommendations with estimated resolution times
- Include both immediate actions and long-term preventive measures

Always structure your analysis with:
- Root Cause: Primary technical cause
- Contributing Factors: Secondary issues that amplified the problem  
- Impact Assessment: Quantified business and technical impact
- Resolution Steps: Specific actions to resolve the issue
- Prevention Measures: Steps to prevent recurrence

Provide confident, decisive recommendations based on the evidence."""

# Initialize the Strands Agent
if bedrock_model:
    sre_agent = Agent(
        model=bedrock_model,
        tools=[query_kubernetes_api, search_application_logs, get_performance_metrics, consult_runbooks],
        system_prompt=sre_system_prompt
    )
    print("🤖 Strands SRE Agent initialized successfully")
    print("   Framework: Strands Agents with Amazon Bedrock")
    print("   Model: Claude 3 Haiku")
    print("   Tools: 4 specialized SRE tools")
    print("   Capabilities: Kubernetes analysis, log search, metrics, runbooks")
else:
    print("⚠️  Could not initialize Strands SRE Agent")
    print("   Please check AWS credentials and Bedrock access")
    sre_agent = None
````

````python
# Run the Strands AI-powered agent investigation
alert_description = "Payment processing service showing 87% error rate and 8.5s response times"

print("🚨 STARTING STRANDS AI INVESTIGATION")
print("═" * 60)
print(f"Alert: {alert_description}")
print(f"AI Framework: Strands Agents")
print(f"AI Model: {BEDROCK_MODEL_ID}")
print()

# Time the investigation
start_time = time.time()

# Run Strands agent investigation
if sre_agent:
    try:
        print("🤖 Strands SRE Agent is analyzing the incident...")
        print("   The agent will automatically use its tools to investigate")
        print("   - query_kubernetes_api()")
        print("   - search_application_logs()") 
        print("   - get_performance_metrics()")
        print("   - consult_runbooks()")
        print()
        
        # Single call to Strands agent - it handles tool orchestration automatically
        investigation_prompt = f"""
        URGENT PRODUCTION INCIDENT: {alert_description}
        
        Please investigate this critical incident immediately. Use all available tools to:
        1. Check Kubernetes pod status and cluster health
        2. Search application logs for error patterns
        3. Review performance metrics and resource usage
        4. Consult relevant operational runbooks
        
        Provide a comprehensive analysis with:
        - Root cause identification with confidence level
        - Contributing factors
        - Business impact assessment  
        - Immediate resolution steps
        - Prevention measures for future incidents
        
        This is affecting customer payments - time is critical!
        """
        
        response = sre_agent(investigation_prompt)
        investigation_result = response.message['content'][0]['text']
        
        end_time = time.time()
        actual_time = round(end_time - start_time, 2)

        print(f"⚡ Strands AI investigation completed in {actual_time} seconds!")
        print("   (vs. 45-65 minutes manually)")
        print("   Agent automatically orchestrated all tools and analysis")
        print()
        
        # Store result for later display
        investigation_report = {
            'timestamp': datetime.now().isoformat(),
            'incident_id': 'INC-20240115-001',
            'framework': 'Strands Agents',
            'ai_model': BEDROCK_MODEL_ID,
            'investigation_time': f"{actual_time} seconds",
            'full_analysis': investigation_result,
            'tools_used': ['query_kubernetes_api', 'search_application_logs', 'get_performance_metrics', 'consult_runbooks']
        }
        
    except Exception as e:
        print(f"❌ Strands AI investigation failed: {e}")
        print("   Please check AWS credentials and Bedrock access")
        
        # Create fallback report for demo continuity
        investigation_report = {
            'incident_id': 'INC-20240115-001',
            'framework': 'Strands Agents (Demo mode - AI unavailable)',
            'ai_model': BEDROCK_MODEL_ID,
            'investigation_time': 'Demo mode',
            'full_analysis': 'AI investigation unavailable - please check AWS credentials',
            'timestamp': datetime.now().isoformat(),
            'tools_used': []
        }
else:
    print("❌ Strands Agent not available")
    print("   Please check AWS credentials and Bedrock model access")
    
    investigation_report = {
        'incident_id': 'INC-20240115-001', 
        'framework': 'Strands Agents (Not initialized)',
        'ai_model': BEDROCK_MODEL_ID,
        'investigation_time': 'N/A',
        'full_analysis': 'Strands Agent could not be initialized',
        'timestamp': datetime.now().isoformat(),
        'tools_used': []
    }
````
